Question: Does a particular product category (e.g., 'Electronics') have significantly higher average sales amounts compared to another category (e.g., 'Clothing')?

H0: The average sales amount for 'Electronics' is equal to the average sales amount for 'Clothing'.

H1: The average sales amount for 'Electronics' is different from the average sales amount for 'Clothing'.

```python
from scipy import stats

# Extract sales amounts for 'Electronics' and 'Clothing' (corrected category name)
electronics_sales = df[df['product_category'] == 'Electronics']['sales_amount']
clothing_sales = df[df['product_category'] == 'Clothing']['sales_amount']

# Perform independent samples t-test
t_statistic, p_value = stats.ttest_ind(electronics_sales, clothing_sales, equal_var=False) # equal_var=False for Welch's t-test, assuming unequal variances

print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpret the results (using a common significance level alpha = 0.05)
alpha = 0.05
if p_value < alpha:
    print(f"\nSince the p-value ({p_value:.4f}) is less than alpha ({alpha}), we reject the null hypothesis.")
    print("Conclusion: There is a statistically significant difference in the average sales amounts between Product categories.")
else:
    print(f"\nSince the p-value ({p_value:.4f}) is greater than alpha ({alpha}), we fail to reject the null hypothesis.")
    print("Conclusion: There is no statistically significant difference in the average sales amounts between Product categories.")
```

Question: Is profit different between different product categories?

H0: There is no significant difference in profit between different product categories.

H1: There is atleast some difference in profit between different product categories.

```python
from scipy import stats

# Get unique product categories
categories = df['product_category'].unique()

# Prepare profit data for each product category
category_profits = []
for category in categories:
    category_profits.append(df[df['product_category'] == category]['profit'])

# Perform one-way ANOVA
f_statistic, p_value = stats.f_oneway(*category_profits)

print(f"F-statistic: {f_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpret the results
alpha = 0.05
if p_value < alpha:
    print(f"\nSince the p-value ({p_value:.4f}) is less than alpha ({alpha}), we reject the null hypothesis.")
    print("Conclusion: There is a statistically significant difference in the average profits between the product categories.")
else:
    print(f"\nSince the p-value ({p_value:.4f}) is greater than alpha ({alpha}), we fail to reject the null hypothesis.")
    print("Conclusion: There is no statistically significant difference in the average profits between the product categories.")
```

Question: Does a particular region has significantly higher sales than other regions?

H0: There is no significant differences in sales between different regions.

H1: There is some significant difference in sales between different regions.

```python
from scipy import stats

# Get unique regions
regions = df['region'].unique()

# Prepare sales data for each region
region_sales = []
for region in regions:
    region_sales.append(df[df['region'] == region]['sales_amount'])

# Perform one-way ANOVA
f_statistic, p_value = stats.f_oneway(*region_sales)

print(f"F-statistic: {f_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpret the results (using a common significance level alpha = 0.05)
alpha = 0.05
if p_value < alpha:
    print(f"\nSince the p-value ({p_value:.4f}) is less than alpha ({alpha}), we reject the null hypothesis.")
    print("Conclusion: There is a statistically significant difference in the average sales amounts between the regions.")
    print("Further post-hoc tests (e.g., Tukey's HSD) would be needed to determine which specific regional pairs differ.")
else:
    print(f"\nSince the p-value ({p_value:.4f}) is greater than alpha ({alpha}), we fail to reject the null hypothesis.")
    print("Conclusion: There is no statistically significant difference in the average sales amounts between the regions.")
```

Question: Does different payment methods has different sales amount associated?

H1: There is atleast some difference between payment method and sales amount

## Question: Is there a significant relationship between `return_flag` and `product_category`?

H0: `return_flag` is independent of `product_category`.

H1: `return_flag` is dependent on `product_category` (i.e., there is a significant relationship).

```python
from scipy.stats import chi2_contingency

# Create a contingency table
contingency_table = pd.crosstab(df['return_flag'], df['product_category'])
display(contingency_table)

# Perform the Chi-square test
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"\nChi-square Statistic: {chi2:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Degrees of Freedom: {dof}")

# Interpret the results (using a common significance level alpha = 0.05)
alpha = 0.05
if p_value < alpha:
    print(f"\nSince the p-value ({p_value:.4f}) is less than alpha ({alpha}), we reject the null hypothesis.")
    print("Conclusion: There is a statistically significant relationship between return_flag and product_category.")
else:
    print(f"\nSince the p-value ({p_value:.4f}) is greater than alpha ({alpha}), we fail to reject the null hypothesis.")
    print("Conclusion: There is no statistically significant relationship between return_flag and product_category.")
```

## Question: Is there a significant relationship between `payment_method` and `region`?

H0: `payment_method` is independent of `region`.

H1: `payment_method` is dependent on `region` (i.e., there is a significant relationship).

```python
from scipy.stats import chi2_contingency

# Create a contingency table for payment_method and region
contingency_table_pr = pd.crosstab(df['payment_method'], df['region'])
display(contingency_table_pr)

# Perform the Chi-square test
chi2_pr, p_value_pr, dof_pr, expected_pr = chi2_contingency(contingency_table_pr)

print(f"\nChi-square Statistic: {chi2_pr:.4f}")
print(f"P-value: {p_value_pr:.4f}")
print(f"Degrees of Freedom: {dof_pr}")

# Interpret the results (using a common significance level alpha = 0.05)
alpha = 0.05
if p_value_pr < alpha:
    print(f"\nSince the p-value ({p_value_pr:.4f}) is less than alpha ({alpha}), we reject the null hypothesis.")
    print("Conclusion: There is a statistically significant relationship between payment method and region.")
    print("This suggests that some payment methods are more preferred in certain regions than others.")
    print("To identify the strongest regional preferences, we can examine the observed vs. expected frequencies or standardized residuals.")
else:
    print(f"\nSince the p-value ({p_value_pr:.4f}) is greater than alpha ({alpha}), we fail to reject the null hypothesis.")
    print("Conclusion: There is no statistically significant relationship between payment method and region.")
```

## Question: What is the linear relationship between `sales_amount` and `unit_price`?

We will use Ordinary Least Squares (OLS) regression to model `sales_amount` as a function of `unit_price` and report key statistics including R², the coefficient, p-value, and 95% confidence intervals for the slope.

```python
import statsmodels.formula.api as smf

# Define the OLS model formula
model_formula = 'sales_amount ~ unit_price'

# Fit the OLS model
ols_model = smf.ols(formula=model_formula, data=df).fit()

# Print the model summary
print(ols_model.summary())

# Extract and report specific values for clarity
r_squared = ols_model.rsquared
unit_price_coeff = ols_model.params['unit_price']
unit_price_pvalue = ols_model.pvalues['unit_price']
unit_price_conf_int = ols_model.conf_int().loc['unit_price']

print(f"\n--- Key Regression Metrics ---")
print(f"R-squared: {r_squared:.4f}")
print(f"Coefficient for unit_price: {unit_price_coeff:.4f}")
print(f"P-value for unit_price: {unit_price_pvalue:.4f}")
print(f"95% Confidence Interval for unit_price slope: ({unit_price_conf_int[0]:.4f}, {unit_price_conf_int[1]:.4f})")
```
